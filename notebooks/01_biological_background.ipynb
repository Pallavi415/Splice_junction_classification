## 1.1 What is DNA?

**DNA (Deoxyribonucleic Acid)** is the molecule that stores genetic information in living organisms.

DNA is made up of smaller units called **nucleotides**. Each nucleotide contains one of four nitrogenous bases:

- **A — Adenine**
- **T — Thymine**
- **G — Guanine**
- **C — Cytosine**

Therefore, a DNA sequence can be represented as a string of these bases.

For example:

```text

## 1.2 What is a Gene?

A **gene** is a region of DNA that contains information used to produce a functional biological product, such as a protein or a functional RNA molecule.

A gene is not simply one continuous block of information. In many genes, the sequence contains regions called **exons** and **introns**.

This distinction becomes important for this project because the dataset focuses on the boundaries between these regions.

## 1.15 Choosing the Value of k

The number of possible k-mers increases rapidly as `k` increases.

Because DNA has four possible nucleotides, the number of possible k-mers is:

**4ᵏ**

For example:

| k | Possible k-mers |
|---:|---:|
| 2 | 16 |
| 3 | 64 |
| 4 | 256 |
| 5 | 1,024 |

A smaller value of `k` captures shorter sequence patterns, while a larger value can capture longer and potentially more specific patterns.

However, larger values also produce more features and may increase computational complexity or the risk of overfitting.

I will therefore investigate suitable k-mer sizes during the feature-engineering stage rather than assuming that one value is automatically the best.
