## Project Outcome

The objective of this project was to classify DNA sequences into three splice-junction classes:

- **EI** - exon to intron boundary
- **IE** - intron to exon boundary
- **N** - neither

I represented each 60-nucleotide sequence using **3-mer frequency features** and used these features to train two classification models: **SVM** and **Random Forest**.

Both models were able to learn meaningful patterns from the DNA sequences, achieving test accuracies above 67%.

## Model Behavior

The models did not perform equally across all three classes.

The **N class** achieved the highest recall for both models. Random Forest achieved a recall of **0.8852**, while SVM achieved **0.8097**.

In contrast, the **IE class** was more difficult to identify. Its recall was **0.4183** for Random Forest and **0.4902** for SVM.

This suggests that the sequence patterns distinguishing splice-junction classes are not equally easy to learn from the available 3-mer representation.

## SVM vs Random Forest

Random Forest achieved a slightly higher overall accuracy than SVM:

- **SVM:** 67.45%
- **Random Forest:** 68.40%

However, SVM achieved a higher macro F1-score:

- **SVM:** 0.6400
- **Random Forest:** 0.6339

This difference is important because macro F1 gives equal importance to all three classes.

Therefore, Random Forest had a small advantage in overall performance, while SVM provided slightly more balanced performance across the classes.

## What the 3-mer Features Captured

The 3-mer representation converted each DNA sequence into numerical features based on the frequency of short nucleotide patterns.

Using `k = 3` allowed the models to capture local sequence information while keeping the feature representation relatively compact.

The results show that these local patterns contain enough information for the models to distinguish between the three splice-junction classes to a meaningful extent.

However, the moderate accuracy also indicates that 3-mer frequencies alone do not capture all the information present in the original DNA sequences. The exact position of a pattern and longer sequence dependencies may also be important for identifying splice-junction boundaries.

## Why EI and IE Are More Difficult

The EI and IE classes represent two different directions of splice-junction boundaries.

Both classes are relatively similar in terms of the biological sequence context they represent, which can make them harder to distinguish than the N class.

The class-wise results support this observation. Both models achieved substantially higher recall for N than for EI and IE.

This means that the models were generally better at recognizing sequences that do not represent a splice junction than identifying the specific type of splice boundary.

## Effect of Hyperparameter Tuning

I also performed controlled hyperparameter tuning for both models.

For SVM:

- Baseline accuracy: **67.45%**
- Tuned accuracy: **67.14%**

For Random Forest:

- Baseline accuracy: **68.40%**
- Tuned accuracy: **67.92%**

In both cases, tuning produced slightly lower test accuracy.

Therefore, I retained the simpler baseline configurations rather than increasing model complexity without a performance improvement.

## Limitations

This project has several limitations.

- The feature representation uses only **3-mer frequencies**, so positional and longer-range sequence information is not directly represented.
- The dataset contains some repeated DNA sequences, including a sequence that appears with different class labels. This can make evaluation more challenging and may introduce ambiguity in the learning problem.
- The dataset is relatively small for a biological classification problem, which limits how broadly the results can be generalized.
- The models were evaluated using a single stratified train-test split, so the reported test performance should be interpreted within this experimental setup.
- Accuracy alone is not sufficient for this multiclass problem, which is why class-wise precision, recall, F1-score, and macro F1 were also considered.
